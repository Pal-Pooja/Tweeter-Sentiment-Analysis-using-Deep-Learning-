```python
#Feature Encoding
from sklearn.preprocessing import LabelEncoder
label=LabelEncoder()
```

```python
data["Sentiments"]=label.fit_transform(data["Sentiments"])
```

```python
data["Sentiments"]
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
token=Tokenizer(oov_token="Nothing")
```

```python
token.fit_on_texts(data["Tweets"])
```

```python
token.word_index
```

```python
token.word_counts
```

```python
token.document_count
```

```python
len(token.word_index)
```

```python
Sequnces=token.texts_to_sequences(data["Tweets"])
```

```python
len(Sequnces[7])
```

```python
from keras.utils import pad_sequences
Sequnces=pad_sequences(Sequnces,padding="pre")
```

```python
Sequnces
```

```python
from tensorflow.keras.layers import SimpleRNN,Dense,Input,Embedding
from tensorflow.keras.models import Sequential
```

```python
x=Sequnces
x
```

```python
from tensorflow.keras.models import Sequential
y=data["Sentiments"].values
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
model=Sequential()
model.add(SimpleRNN(32,input_shape=(99,1),return_sequences=False))
model.add(Dense(3,activation="softmax"))
```

```python
model.summary()
```

```python
model.compile(optimizer="adam",metrics=["accuracy"],loss='sparse_categorical_crossentropy')
```

```python
model.fit(x_train,y_train,epochs=10)
```

```python
model.fit(x_train,y_train,epochs=10,validation_data=(x_test,y_test),batch_size=64)
```

```python
vocab_size=len(token.word_index)+1
vocab_size
```

```python
model2=Sequential()
model2.add(Input(shape=(99,)))
model2.add(Embedding(input_dim=vocab_size,output_dim=64))
model2.add(Dense(3,activation="softmax"))
model2.summary()
```

```python
model_final=Sequential([Embedding(input_dim=vocab_size,output_dim=128,input_length=99,mask_zero=True),
SimpleRNN(32),Dense(3,activation="softmax")])
```

```python
model_final.compile(optimizer="adam",metrics=["accuracy"],loss="sparse_categorical_crossentropy")
```

```python
model_final.summary()
```

```python
model_final.fit(x_train,y_train,epochs=100,validation_data=(x_test,y_test),batch_size=64)
```
